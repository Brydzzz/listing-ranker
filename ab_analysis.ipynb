# A/B Test Results Analysis

```python
import polars as pl

df = pl.read_csv("ab_results.csv")
print(f"Total rows: {df.height}")
print(f"Unique sessions: {df['session_id'].n_unique()}")
df.head()
```

## Average Session Length per Model


```python
viewed = df.filter(pl.col("session_type") != "not_viewed")

session_lengths = (
    viewed
    .group_by("session_id", "ranking_model")
    .agg(pl.len().alias("session_length"))
)

avg_by_model = (
    session_lengths
    .group_by("ranking_model")
    .agg(
        pl.col("session_length").mean().alias("avg_session_length"),
        pl.col("session_length").std().alias("std_session_length"),
        pl.col("session_length").median().alias("median_session_length"),
        pl.len().alias("num_sessions"),
    )
    .sort("ranking_model")
)

avg_by_model
```
